# Predicting Flight Delays at Seattle-Tacoma International Airport

**HCDE 410 Final Project Proposal**
Sam Carruthers · Autumn 2026

## 1. Motivation and problem statement

Flight delays are one of the most common frustrations in travel, and their costs land on people unevenly. A one-hour delay is an annoyance for some travelers. For others it means a missed connection, a lost shift, or a stranded caregiver. Airlines and airports publish average on-time rates, but an average can't tell a traveler how risky a specific flight is when they are choosing between options.

This project will use machine learning to predict whether a domestic flight departing Seattle-Tacoma International Airport (SEA) will arrive at its destination 15 or more minutes late. The model will use only information a traveler could know at booking time: airline, scheduled departure time, day of week, month, destination, and distance.

**Why it is interesting and useful**

Part of the question is scientific. It tests how much of flight delay is predictable from schedule features alone and how much comes from things no schedule can anticipate, like weather, air traffic control, and knock-on delays from earlier flights. That answer is useful even if the model is only moderately accurate. It is also practical: if certain departure windows, routes, or carriers carry consistently higher delay risk, travelers can factor that into booking decisions, especially for tight connections.

There is a human-centered side too. Predictions shown to travelers need to be interpretable and honestly calibrated, because a model that sounds confident and turns out wrong shifts risk onto the people trusting it. So besides accuracy, I want to check whether the model's probabilities mean what they say.

**What I hope to learn**

I plan to train two simple, interpretable classification models, logistic regression and a shallow decision tree, and compare them against a naive baseline that always predicts "on time." I will evaluate them with a confusion matrix, precision and recall on the delayed class, and a simple calibration check, and I will use the logistic regression coefficients and the tree's splits to see which factors drive predicted delay risk. Overall, I want to learn which choices within a traveler's control matter most, how much delay risk can't be predicted from a schedule, and how to communicate a delay prediction without overstating it.

## 2. Data

**Dataset:** Reporting Carrier On-Time Performance (1987 to present), Airline On-Time Performance Data, U.S. Bureau of Transportation Statistics (BTS), TranStats.

- Database overview: https://transtats.bts.gov/DatabaseInfo.asp?QO_VQ=EFD
- Download page: https://www.transtats.bts.gov/DL_SelectFields.asp?gnoyr_VQ=FGJ&QO_fu146_anzr=

**What it represents**

Each row is one scheduled domestic flight operated by a U.S. carrier that accounts for at least 0.5% of domestic scheduled passenger revenue. Fields include:

- Flight date, reporting airline, flight number, and tail number
- Origin and destination airports, and distance
- Scheduled and actual departure and arrival times
- Departure and arrival delay in minutes, plus a 15-minute arrival delay indicator (`ArrDel15`), which will be my prediction target
- Cancellation and diversion indicators, with cancellation reason codes
- Minutes of delay attributed to each cause: carrier, weather, National Airspace System, security, and late-arriving aircraft (available since June 2003)

Data is released monthly, roughly three months behind; the latest month currently available is August 2026. I plan to filter to flights departing SEA from January 2023 through the most recent month, which keeps the data large enough for machine learning but small enough to work with in a notebook.

**License and terms of use**

The data is published by a U.S. federal agency. Works of the U.S. federal government are not subject to copyright (17 U.S.C. § 105), so the data is in the public domain. BTS does not list additional usage restrictions on the database pages; I will cite BTS as the source.

**Why this dataset fits**

It provides a real outcome label for every flight, a large sample, and schedule features that are known before a flight departs, which lets me build a model that matches what a traveler would actually know. The delay cause fields are not usable as model inputs (they are only known after the fact), but they will help me interpret where the model succeeds and fails.

**Ethical considerations**

- **No passenger data:** The dataset describes flights, not people, so there is no direct privacy risk to travelers. Tail and flight numbers could be linked to specific crews in principle, but I will not use them to identify individuals.
- **Self-reported causes:** Airlines assign delay causes themselves, and categories like "carrier" versus "National Airspace System" can be misattributed in ways that favor the reporting airline. I will treat cause data as context, not ground truth.
- **Coverage gaps:** Only carriers above the revenue threshold report, so smaller airlines are missing and the model will not represent every SEA traveler's options.
- **Cancellations:** Cancelled flights have no arrival delay. Dropping them would hide the worst outcomes and make some flights look more reliable than they are, so I will report them separately rather than silently excluding them.
- **Misuse and overconfidence:** Delay rankings could be read as definitive verdicts on airlines, or reused in ways that shift costs onto travelers (for example, risk-based pricing). I will present predictions with their uncertainty and calibration and avoid framing results as airline rankings.
- **Who bears the harm is not in the data:** The dataset cannot show which travelers are hurt most by delays. I will acknowledge this limit rather than make claims about impacts I cannot measure.

## 3. Unknowns and dependencies

- **Data availability:** TranStats is a federal website that can be slow or temporarily unavailable, including during a lapse in federal funding. I will download the full date range early and keep a local copy so the project does not depend on the site later in the quarter.
- **Download format:** TranStats exports one month per download, so assembling multiple years requires scripting or many manual downloads. I will budget time for this in the first week.
- **Model performance:** Delays are driven partly by events schedule features cannot capture, so accuracy may be modest. That is still a meaningful finding, but it could limit how much the coefficients and tree splits can say about what drives delays.
- **Scope:** Adding outside data such as hourly weather from NOAA would likely improve predictions but would also add joining and cleaning work. I will treat it as a stretch goal only if the core model is finished.
- **Time:** I am balancing this project with other courses that have end-of-quarter deadlines, so I will keep the core analysis small and add depth only after it works end to end.
